```python
# 1. Import Libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline

# 2. Load Data
iris = sns.load_dataset('iris')
print("First 5 rows of dataset:")
print(iris.head())

# 3. Exploratory Data Analysis
print("\nGenerating Pairplot...")
sns.pairplot(iris, hue='species')
plt.show()

print("\nKDE Plot for Setosa...")
setosa = iris[iris['species'] == 'setosa']
sns.kdeplot(x=setosa['sepal_length'], y=setosa['sepal_width'], cmap='coolwarm', fill=True)
plt.show()

# 4. Train Test Split
from sklearn.model_selection import train_test_split

X = iris.drop('species', axis=1)
y = iris['species']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)

# 5. Train SVM Model
from sklearn.svm import SVC

model = SVC()
model.fit(X_train, y_train)

# 6. Model Evaluation
from sklearn.metrics import classification_report, confusion_matrix

predictions = model.predict(X_test)

print("\nConfusion Matrix (Before Tuning):")
print(confusion_matrix(y_test, predictions))

print("\nClassification Report (Before Tuning):")
print(classification_report(y_test, predictions))

# 7. Grid Search (Hyperparameter Tuning)
from sklearn.model_selection import GridSearchCV

param_grid = {
    'C': [0.1, 1, 10, 100],
    'gamma': [1, 0.1, 0.01, 0.001]
}

print("\nRunning GridSearchCV...")
grid = GridSearchCV(SVC(), param_grid, refit=True, verbose=2)
grid.fit(X_train, y_train)

print("\nBest Parameters Found:")
print(grid.best_params_)

# 8. Evaluation After Tuning
grid_predictions = grid.predict(X_test)

print("\nConfusion Matrix (After Tuning):")
print(confusion_matrix(y_test, grid_predictions))

print("\nClassification Report (After Tuning):")
print(classification_report(y_test, grid_predictions))
```
